```python
import numpy as np
import csv
#csv file with radii, radii_err, star, TIC_ID, spec_type, star_mass_sol, star_lum_sol, star_distance_pc, telescope,
# from Lecavelier des Etangs, A., Cros, L.,H´ebrard, G., et al. 2022, Scientific Reports, 12, 5855, doi: 10.1038/s41598-022-09021-2
# and sources cited within the text
# used webplotdigitizer 
star = 'RZ Psc'
TIC_ID = 15693497
spec_type = 'K'
star_mass_sol = 1.1
star_rad_sol = 1.0
star_lum_sol = 0.8
star_distance_pc = 184.1
telescope = 'TESS'
point1_a = np.array([0.7765676631607591, 0.4015964849042396]) # x1, y1
point2_a = np.array([2.5249706588372396, 0.23558557379674408]) # x2, y2
# split distribution
point1_b = np.array([2.6931930571359044, 0.20978711676762363]) # x1, y1
point2_b = np.array([7.26794122855926, 0.02279988707609292]) # x2, y2
filename = '/Users/adalyngibson/Documents/Github/MidtermProject/data/midtermdata.csv'
with open(filename, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow([
        "x1", "y1", "x2", "y2",
        "star", "TIC_ID", "spec_type",
        "star_mass_sol", "star_rad_sol", "star_lum_sol", "star_distance_pc", "telescope"
    ])

    for p1, p2 in zip((point1_a, point1_b), (point2_a, point2_b)):
        writer.writerow([
            float(p1[0]), float(p1[1]),
            float(p2[0]), float(p2[1]),
            star, TIC_ID, spec_type,
            float(star_mass_sol), float(star_rad_sol), float(star_lum_sol),
            float(star_distance_pc), telescope
        ])
```
